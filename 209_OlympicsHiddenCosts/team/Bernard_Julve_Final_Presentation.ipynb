## Attempt 1

```
# Do some clean up based on the missingness analysis above.

# Remove unsupported types on 'Domestic Sponsorship Revenue (USD 2018)'. Since it specifically
# says No domestic sponshorship, we are assuming the value can be accurately represented as 0,
# as oppose to unreported.
df_olympics['Domestic Sponsorship Revenue (USD 2018)'] = df_olympics['Domestic Sponsorship Revenue (USD 2018)'].replace({'No domestic sponsorship': 0}).infer_objects(copy=False)
print(df_olympics['Domestic Sponsorship Revenue (USD 2018)'].unique(), '\n')
```

```
# Start with a chart plotting the total revenue vs the total cost.
import altair as alt

# Need to create two new columns for the totals.
df_olympics['Revenue (USD2018)'] = df_olympics['Ticketing Revenue (USD2018)'] + df_olympics['Broadcast Revenue (USD2018)'] + df_olympics['International Sponsorship Revenue (USD 2018)'] + df_olympics['Domestic Sponsorship Revenue (USD 2018)']
df_olympics['Cost (USD2018)'] = df_olympics['Cost of venues (USD 2018)'] + df_olympics['Cost of organisation (USD 2018)']

# Chart as two connected line graphs.
chart = alt.Chart(df_olympics).transform_fold(
    ['Revenue (USD2018)', 'Cost (USD2018)'],
    as_=['Total', 'Amount (USD2018)']
).mark_line().encode(
    x='Year:Q',
    y='Amount (USD2018):Q',
    color='Total:N'
).properties(
    title='Total Revenue vs Total Cost for Olympic Hosts 1964-2024',
    width=600,
    height=400
)

chart.display()
```

**What's informative about this view:** This view shows the change in total revenue vs total cost year over year. It is helpful in showing the general trends in both categories, with the total cost consistently higher than the total revenues being reported in each Olympic event year.

**What could be improved about this view:**  It requires a bit of work for the user to repeatedly switch between each line to do the comparison year to year. There are also some years where the revenue exceeds or is about equal to the costs, but it's not immediately clear until you scan each X axis interval. Also, the X axis is a bit misleading, as the data reported is actually only for even numbered years which are Olympic years, whereas on the graph, the line chart implies a continuity over the timespan which is not necessarily the case.


## Attempt 2

```
chart = alt.Chart(df_olympics).mark_line(point=True).encode(
    x='Revenue (USD2018):Q',
    y='Cost (USD2018):Q',
    order='Year:Q',
    tooltip=['Year:Q', 'Revenue (USD2018):Q', 'Cost (USD2018):Q']
).properties(
    width=600,
    height=400,
    title='Total Revenue vs Total Cost for Olympic Hosts 1964-2024'
)

chart.display()
```

**What's informative about this view:** This view shows the change in total revenue vs total cost without relying on an X axis time scale. It removes the implication that odd year information is included in the line interpolation, and thus is a more accurate representation of the data, preserving expressiveness.

**What could be improved about this view:**  Due to the high variance of the data, the graph is too crowded to be able to distinguish the points. At some places the points are bunched together, and at times the graph crosses over itself repeatedly. In addition, it's not clear which points come earlier and which points come later in the time series, as the line is missing an arrow for direction. Thus, it is difficult for the user to determine the changes in revenue and cost year to year.

```
# Reformat dataset for the streamgraph.
# Started by Google searching for streamgraph examples in Altair and
# modifying to our needs.
alt.Chart(df_olympics).transform_fold(
    ['Revenue (USD2018)', 'Cost (USD2018)'],
    as_=['Total', 'Value (USD2018)']
).mark_area().encode(
    alt.X('Year:Q'),
    alt.Y('Value (USD2018):Q').stack('center').axis(None),
    alt.Color('Total:N'),
    tooltip=['Year:Q', 'Value (USD2018):Q', 'Country']
).properties(
    width=600,
    height=400,
    title='Total Revenue vs Total Cost for Olympic Hosts 1964-2024'
).interactive()
```

**What's informative about this view:** This view shows the change in total revenue vs total cost, relying on area instead of lines to summarize the difference in totals. This also might be a bit misleading regarding continuity similar to the line graph, but the emphasis instead is in the shaded areas which might serve to shift the user focus, as the difference in totals can be deduced from the difference in total area of each colored section rather than relying on the user to compare the exact values represented by different lines. Thus, the task for the user is visually immediate for these particular variables from this dataset. Also, hover tooltips were added to list explicitly the Total value for each part of the graph, as well as interactivity was added to allow the user to zoom in on particular areas.

**What could be improved about this view:**  It would be helpful to add in dotted lines to indicate exact Olympic years without having to hover on the area. Similarly, adding some country flags to the graph to denote the host nation would also be helpful.

```
# Experimented with different colors.
df_olympics['Year'] = pd.to_datetime(df_olympics['Year'], format='%Y', errors='coerce')

year_selector = alt.selection_point(fields=['Year'], name='selected_year')

streamgraph_costs_revenues = alt.Chart(df_olympics).transform_fold(
    ['Revenue (USD2018)', 'Cost (USD2018)'],
    as_=['Total', 'Amount (USD2018)']
).mark_area().encode(
    alt.X(
        'Year:T',
        axis=alt.Axis(
            tickCount={"interval": "year", "step": 2},
            format='%Y'
        )),
    alt.Y('Amount (USD2018):Q').stack('center'),
    alt.Color('Total:N', scale=alt.Scale(
        domain=['Revenue (USD2018)', 'Cost (USD2018)'],
        range=['#2ca02c', '#d62728']  # Green for up/Revenue, Red for down/Cost
    ), legend=None),
    order=alt.Order('Total:N', sort='ascending'),
    opacity=alt.condition(year_selector, alt.value(1.0), alt.value(0.25)),
    tooltip=[alt.Tooltip('Year:T', title='Year', format='%Y'), 'Amount (USD2018):Q', 'Country']
).add_params(
    year_selector
).properties(
    width=400,
    height=400,
    title='Total Revenue vs Total Cost for Olympic Hosts 1964-2024'
)

# Faceted bar chart.
df_olympics_melted = pd.melt(
    df_olympics,
    id_vars=['Cycle', 'Country', 'Year', 'Event'],
    value_vars=['Ticketing Revenue (USD2018)', 'Broadcast Revenue (USD2018)', 'International Sponsorship Revenue (USD 2018)',
                'Domestic Sponsorship Revenue (USD 2018)', 'Cost of venues (USD 2018)', 'Cost of organisation (USD 2018)'], # The 6 columns to combine
    var_name='Category',
    value_name='Amount'
)

financial_type = {
    'Ticketing Revenue (USD2018)': 'Revenue', 'Broadcast Revenue (USD2018)': 'Revenue', 'International Sponsorship Revenue (USD 2018)': 'Revenue',
    'Domestic Sponsorship Revenue (USD 2018)': 'Revenue', 'Cost of venues (USD 2018)': 'Cost', 'Cost of organisation (USD 2018)': 'Cost'
}

df_olympics_melted['Type'] = df_olympics_melted['Category'].map(financial_type)
df_olympics_melted['Category'] = df_olympics_melted['Category'].str.replace(r' \(USD\s?2018\)', '', regex=True)

chart_costs_breakdown = alt.Chart(df_olympics_melted).mark_bar().encode(
    x=alt.X('Category:N', sort='y', axis=alt.Axis(labelAngle=45, title="Financial Categories")),
    y=alt.Y('Amount:Q', axis=alt.Axis(title="Amount (USD 2018)")),
    color=alt.Color('Type:N', scale=alt.Scale(
        domain=['Revenue', 'Cost'],
        range=['#2ca02c', '#d62728']  # Green for up/Revenue, Red for down/Cost
    ))
).transform_filter(
    year_selector
).properties(
    width=200,
    height=400,
    title=alt.TitleParams(
        text=alt.expr("selected_year.Year ? 'Revenue vs. Cost Breakdown for ' + selected_year.Year : 'Revenue vs. Cost Breakdown (All Years)'")
    )
)

midpoint_line = alt.Chart().mark_rule(color='black', strokeWidth=1.5).encode(y=alt.datum(0))
layered_bar_chart = (chart_costs_breakdown + midpoint_line)

dashboard = alt.hconcat(
    streamgraph_costs_revenues,
    layered_bar_chart
).resolve_scale(
    color='independent'
)

dashboard.display()
```

```
# Removed white space gaps from the streamgraph.
# Used Google search AI mode to debug:
# "altair streamgraph gaps"
# It is primarly due to duplicate records for the same year causing the gaps.
# There were some instances where the Summer and Winter Olmpic Games occurred
# in the same year. The solution is to aggregate those records, and in this
# case show those events together (e.g. 1984 Los Angeles/1984 Sarajevo).
df_olympics['Year'] = pd.to_datetime(df_olympics['Year'], format='%Y', errors='coerce')
df_olympics['Year_int'] = df_olympics['Year'].dt.year

df_olympics_aggregated_by_year = df_olympics.groupby('Year_int').agg({
    'Ticketing Revenue (USD2018)': 'sum',
    'Broadcast Revenue (USD2018)': 'sum',
    'International Sponsorship Revenue (USD 2018)': 'sum',
    'Domestic Sponsorship Revenue (USD 2018)': 'sum',
    'Cost of venues (USD 2018)': 'sum',
    'Cost of organisation (USD 2018)': 'sum',
    'Revenue (USD2018)': 'sum',
    'Cost (USD2018)': 'sum',
    'Cycle': lambda x: '/'.join(x.dropna().unique()),
    'Country': lambda x: '/'.join(x.dropna().unique()),
    'Event': lambda x: '/'.join(x.dropna().unique()),
    'Year': 'first'
}).reset_index()

# Format dollar amounts.
def format_dollars(val):
    if abs(val) >= 1_000_000_000:
        return f"${val / 1_000_000_000:,.2f} B"
    elif abs(val) >= 1_000_000:
        return f"${val / 1_000_000:,.2f} M"
    elif abs(val) >= 1_000:
        return f"${val / 1_000:,.1f} K"
    return f"${val:,.2f}"

df_olympics_aggregated_by_year['Revenue'] = df_olympics_aggregated_by_year['Revenue (USD2018)'].apply(format_dollars)
df_olympics_aggregated_by_year['Cost'] = df_olympics_aggregated_by_year['Cost (USD2018)'].apply(format_dollars)
df_olympics_aggregated_by_year['Event'] = df_olympics_aggregated_by_year['Event'].str.replace(r'\(.*\)', '', regex=True)

year_selector = alt.selection_point(
    fields=['Year_int', 'Revenue', 'Cost'],
    name='selected_year',
    nearest=True,
    value=[{'Year_int': 2014, 'Revenue': '$3.20 B', 'Cost': '$15.08 B'}],
    empty=False,
    on='pointerover'
)

# Reusable streamgraph base.
base_stream = alt.Chart(df_olympics_aggregated_by_year).transform_fold(
    ['Revenue (USD2018)', 'Cost (USD2018)'],
    as_=['Total', 'Amount (USD2018)']
).transform_joinaggregate(
    GroupedAmount='sum(Amount (USD2018))',
    groupby=['Year_int', 'Total']
).encode(
    alt.X('Year_int:O', axis=alt.Axis(title="Year")),
    alt.Y('Amount (USD2018):Q',
          axis=None
          ).stack('center'),
    alt.Color('Total:N', scale=alt.Scale(
        domain=['Revenue (USD2018)', 'Cost (USD2018)'],
        #range=['#2ca02c', '#d62728']
    ), legend=None),
    order=alt.Order('Total:N', sort='ascending')
)

# Render main area visuals responding to the selector opacity change.
area_layer = base_stream.mark_area(
    stroke='true',
    strokeWidth=0.5,
    line=False
).encode(
    #opacity=alt.condition(year_selector, alt.value(1.0), alt.value(0.25))
)

# Layer A: The thin, visible indicator line that appears on selection.
visible_indicator = alt.Chart(df_olympics_aggregated_by_year).mark_rule(
    color='#333333',
    strokeWidth=2,
    strokeDash=[4, 4]
).encode(
    alt.X('Year_int:O', axis=alt.Axis(title="Year")),
    opacity=alt.condition(year_selector, alt.value(1.0), alt.value(0.0))
)

# Layer B: The wide, invisible collision boundary for mouse tracking.
invisible_hover_zone = alt.Chart(df_olympics_aggregated_by_year).mark_rule(
    strokeWidth=30,      # Huge invisible hit-box for easy tracking.
    opacity=0.001
).transform_fold(
    ['Revenue (USD2018)', 'Cost (USD2018)'],
    as_=['Total', 'Amount (USD2018)']
).encode(
    alt.X('Year_int:O', axis=alt.Axis(title="Year")),
    tooltip=['Cycle', 'Event', 'Country', 'Revenue', 'Cost']
).add_params(
    year_selector        # Drives the selection state variables.
)

# Layer components.
streamgraph_costs_revenues = alt.layer(
    area_layer,
    visible_indicator,
    invisible_hover_zone
).properties(
    width=400,
    height=400,
    title=alt.TitleParams(
        text='Total Revenue vs Total Cost for Olympic Hosts 1964-2024',
        fontSize=16,
        fontWeight='bold',
        offset=15
    )
)


# Faceted bar chart.
df_olympics_melted = pd.melt(
    df_olympics_aggregated_by_year,
    id_vars=['Cycle', 'Country', 'Year', 'Year_int', 'Event', 'Revenue', 'Cost'],
    value_vars=['Ticketing Revenue (USD2018)', 'Broadcast Revenue (USD2018)', 'International Sponsorship Revenue (USD 2018)',
                'Domestic Sponsorship Revenue (USD 2018)', 'Cost of venues (USD 2018)', 'Cost of organisation (USD 2018)'],
    var_name='Category',
    value_name='Amount'
)

financial_type = {
    'Ticketing Revenue (USD2018)': 'Revenue', 'Broadcast Revenue (USD2018)': 'Revenue', 'International Sponsorship Revenue (USD 2018)': 'Revenue',
    'Domestic Sponsorship Revenue (USD 2018)': 'Revenue', 'Cost of venues (USD 2018)': 'Cost', 'Cost of organisation (USD 2018)': 'Cost'
}

df_olympics_melted['Type'] = df_olympics_melted['Category'].map(financial_type)
df_olympics_melted['Category'] = df_olympics_melted['Category'].str.replace(r' \(USD\s?2018\)', '', regex=True)

chart_costs_breakdown = alt.Chart(df_olympics_melted).mark_bar().encode(
    x=alt.X('Category:N', sort=alt.SortField('Type', 'descending'), axis=alt.Axis(labelAngle=45, title="")),
    y=alt.Y('sum(Amount):Q', axis=alt.Axis(title="$ Amount (USD 2018)")),
    color=alt.Color('Type:N', scale=alt.Scale(
        domain=['Revenue', 'Cost'],
        #range=['#2ca02c', '#d62728']  # Green for up/Revenue, Red for down/Cost
    ))
).transform_filter(
    year_selector
).properties(
    width=200,
    height=400,
    title=alt.TitleParams(
        text=alt.expr("selected_year.Year_int ? 'Totals Breakdown ' + selected_year.Year_int + ' (' + selected_year.Revenue + ' Revenue vs. ' + selected_year.Cost + ' Cost)' : '(Hover over a year to display)'"),
        fontSize=16,
        fontWeight='bold',
        offset=40.25
    )
)

midpoint_line = alt.Chart().mark_rule(color='black', strokeWidth=1.5).encode(y=alt.datum(0))
layered_bar_chart = (chart_costs_breakdown + midpoint_line)

dashboard = alt.hconcat(
    streamgraph_costs_revenues,
    layered_bar_chart
).resolve_scale(
    color='independent'
)

dashboard.display()
```

```
olympic_years = {
    1964: ["Japan", "Austria"],
    1968: ["Mexico", "France"],
    1972: ["Germany", "Japan"],
    1976: ["Canada", "Austria"],
    1980: ["Russian Federation", "United States"],
    1984: ["United States", "Yugoslavia"],
    1988: ["South Korea", "Canada"],
    1992: ["Spain", "France"],
    1994: ["Norway"],
    1996: ["United States"],
    1998: ["Japan"],
    2000: ["Australia"],
    2002: ["United States"],
    2004: ["Greece"],
    2006: ["Italy"],
    2008: ["China, People's Republic of"],
    2010: ["Canada"],
    2012: ["United Kingdom"],
    2014: ["Russian Federation"],
    2016: ["Brazil"],
    2018: ["Korea, Republic of"],
    2020: ["Japan"],
    2022: ["China"],
    2024: ["France"],
    2026: ["Italy"]
}

def generate_olympic_year_filter():
  filter = ''
  for k,v in olympic_years.items():
    for c in v:
      filter += f'(year(datum.Year) === {k} && datum.Country === "{c}") || '
  filter = filter[:-3]
  return filter

olympic_filter = generate_olympic_year_filter()
olympic_filter
```

  - Determine if Olympic event had any wider economic impact on the host nation.

  - Need to compare values between nations.
  - Need to clearly see the Olympic year within the trend.

  - Prospective Voters.
  - Policy Makers.
- Data
  - Public Finances in Modern History (https://www.imf.org/external/datamapper/datasets/FPP).
- What?
  - Tables
  - Attributes:
    - Categorical
    - Quantitative
    - Sequential
- Why?
  - Action:
    - Discover
    - Compare
  - Target:
    - Trends
    - Features
- Design Choices:
  - Indexed line chart most scalable for multiple host nations.
  - Hover tooltips to display more detailed information across all selected host nations.
  - Allow user to select particular host nations to declutter the visualization.
  - Add icons to denote summer vs winter Olympic event.

```
import altair as alt

# Line chart first iteration. Used the Primary Balance dataset for EDA. Prompt:
# "altair multi select"
selection = alt.selection_point(fields=['Country'], bind='legend', name="highlight_select")
dataset_with_name = alt.Data(values=df_olympics_balance_gdp.to_dict(orient='records'), name="df_olympics_balance_gdp")

cursor_param = alt.param(
    name='cursor',
    value='default',
    expr=f"data('{selection.name}') ? 'pointer' : 'default'"
)

hover = alt.selection_point(
    fields=['Year'], nearest=True, on='mouseover', empty='none', clear='mouseout'
)

lines = alt.Chart(dataset_with_name).mark_line().encode(
    x=alt.X(
        'Year:T',
        axis=alt.Axis(
            tickCount='year',
            tickMinStep=2,
            format='%Y'
        )),
    y='Primary Balance GDP Percentage:Q',
    color='Country:N',
    opacity=alt.condition(selection, alt.value(1.0), alt.value(0.1))
).add_params(
    selection,
    cursor_param
).properties(
    width=600,
    height=500,
    title='Primary Balance GDP Percentage for Olympic Hosts 1964-2024'
)

tooltips = alt.Chart(dataset_with_name).transform_pivot(
    'Country', value='Primary Balance GDP Percentage', groupby=['Year']
).mark_rule(color='lightgray').encode(
    x='Year:T',
    tooltip=[alt.Tooltip('Year:T', title='Year', format='%Y')] + [alt.Tooltip(c.replace("'", "\\'"), type='quantitative') for c in list(df_olympics_balance_gdp['Country'].unique())]
).add_params(hover)

# Layer: Indicate the Summer and Winter Olympic years on each trend line.
dots = alt.Chart(dataset_with_name).transform_filter(
    # Olympic Years Only
    olympic_filter
).mark_point(
    size=250,
    filled=True,
    opacity=1,
    shape='diamond'
).encode(
    x='Year:T',
    y='Primary Balance GDP Percentage:Q',
    color='Country:N',
    opacity=alt.condition(selection, alt.value(1.0), alt.value(0.1))
)

(tooltips + lines + dots).display()
```

```
import altair as alt
import numpy as np

# The final version, incorporating all user feedback including limiting to
# post 1970 records for consistency with Hypothesis 4 charts, reformatting
# dollar amounts for easier readability, and allowing (fixing) multi select.
# Also, instead of proceeding with the filtered tooltip/label text approach,
# the logic was changed to only show that information when hovering the summer
# or winter dots. This is when that information is most relevant and needed
# anyway, and so made sense to show at that point rather than always showing
# regardless of where the user hovered. Also, besides showing the GDP growth rate
# for that Olympic year, a Would Have was added where the average GDP for a
# certain time window before and after the Olympic year was added to the tooltip,
# allowing the user to view summary statistics to help them decide if there was
# a correlation between the Olympic event and before/after GDP growth rates.
# Prompts:
# "pandas calculate averages 5 years before and 10 years after window"


# Main Legend Selection.
# This object owns the 'highlight_select' parameter name in Vega.
selection = alt.selection_point(
    fields=['Country'],
    bind='legend',
    name="highlight_select",
    value=[{'Country': 'Korea, Republic of'}, {'Country': 'United States'}],
    on='click',
    toggle=True
)

df_olympics_balance_gdp_post_1970 = df_olympics_balance_gdp[df_olympics_balance_gdp['Year'] >= '1970-01-01']
df_olympics_balance_gdp_post_1970 = df_olympics_balance_gdp[df_olympics_balance_gdp['Country'] != 'Mexico']

# Format dollar amounts.
def format_dollars(val):
    if abs(val) >= 1_000_000_000:
        return f"${val / 1_000_000_000:,.2f} B"
    elif abs(val) >= 1_000_000:
        return f"${val / 1_000_000:,.2f} M"
    elif abs(val) >= 1_000:
        return f"${val / 1_000:,.1f} K"
    return f"${val:,.2f}"

# Map Total Revenues and Total Costs safely using .loc.
df_olympics_balance_gdp_post_1970['TotalRevenue'] = df_olympics_balance_gdp_post_1970['Revenue (USD2018)'].apply(format_dollars)
df_olympics_balance_gdp_post_1970['TotalCost'] = df_olympics_balance_gdp_post_1970['Cost (USD2018)'].apply(format_dollars)

# Safely strip text/symbols and cast to a brand new numeric float column.
# This prevents modifying your original 'GDP Growth Percentage' string data column.
clean_gdp = df_olympics_balance_gdp_post_1970['GDP Growth Percentage'].astype(str).str.replace('%', '', regex=False).str.strip()
df_olympics_balance_gdp_post_1970['GDP Growth Percentage Numeric'] = pd.to_numeric(clean_gdp, errors='coerce')

# Initialize placeholders for the new metrics
df_olympics_balance_gdp_post_1970.loc[:, 'Total Revenues'] = np.nan
df_olympics_balance_gdp_post_1970.loc[:, 'Total Costs'] = np.nan
df_olympics_balance_gdp_post_1970.loc[:, 'Avg GDP (>=5 Yrs Before)'] = np.nan
df_olympics_balance_gdp_post_1970.loc[:, 'Avg GDP (>=10 Yrs After)'] = np.nan

# Extract rows that hosted an Olympic Game
host_years_df = df_olympics_balance_gdp_post_1970[
    (df_olympics_balance_gdp_post_1970['Year'].dt.year.isin([year['year'] for year in summer_olympics + winter_olympics]))
]

# Compute rolling time horizons per host country event
for idx, row in host_years_df.iterrows():
    country = row['Country']
    olympic_year = row['Year']

    # Isolate all data rows belonging to this specific country
    country_data = df_olympics_balance_gdp_post_1970[df_olympics_balance_gdp_post_1970['Country'] == country]

    # Calculate relative distance in years from the Olympic year
    # Negative numbers mean BEFORE the Olympics, positive numbers mean AFTER
    years_diff = (country_data['Year'] - olympic_year).dt.days / 365.25

    # Filter conditions for your revised horizons
    # '5 years or more before': values like -5, -6, -10 (years_diff <= -5)
    mask_5_years_before = (years_diff <= -5)
    # '10 years or more after': values like 10, 11, 15 (years_diff >= 10)
    mask_10_years_after = (years_diff >= 10)

    # Calculate localized averages
    avg_gdp_before = country_data.loc[mask_5_years_before, 'GDP Growth Percentage Numeric'].mean()
    avg_gdp_after = country_data.loc[mask_10_years_after, 'GDP Growth Percentage Numeric'].mean()

    # Map values back to the specific Olympic event index row
    df_olympics_balance_gdp_post_1970.at[idx, 'Avg GDP (>=5 Yrs Before)'] = avg_gdp_before
    df_olympics_balance_gdp_post_1970.at[idx, 'Avg GDP (>=10 Yrs After)'] = avg_gdp_after


dataset_with_name = alt.Data(
    values=df_olympics_balance_gdp_post_1970.to_dict(orient='records'),
    name="df_olympics_balance_gdp_post_1970"
)

cursor_param = alt.param(
    name='cursor',
    value='default',
    expr=f"data('highlight_select') ? 'pointer' : 'default'"
)

# Adding point='transparent' creates invisible, highly clickable hover hitboxes
# over every line vertex natively without breaking layered components.
lines = alt.Chart(dataset_with_name).mark_line().encode(
    x=alt.X(
        'Year:T',
        axis=alt.Axis(
            tickCount='year',
            tickMinStep=2,
            format='%Y'
        )),
    y='GDP Growth Percentage:Q',
    color=alt.Color(
        'Country:N',
        legend=alt.Legend(
            title="Click Countries to Select",
            symbolSize=20
        )),
    opacity=alt.condition(selection, alt.value(1.0), alt.value(0.1)),
    strokeWidth=alt.condition(selection, alt.value(2.5), alt.value(1))
).add_params(
    selection,
    cursor_param
).properties(
    width='container',
    height=500,
    title=alt.TitleParams(
        text='GDP Growth Percentage for Olympic Hosts 1970-2024',
        # subtitle='Click on a country name in the legend to view its line. Hold SHIFT and click to select multiple countries for comparison. Olympic Cycle not selectable.',
        # subtitleColor='gray',
        # subtitleFontSize=12
    )
)

# This dummy chart evaluates first to force the Olympic Cycle legend to the top
# of the sidebar column without interrupting the lines interaction registry.
cycle_legend_anchor = alt.Chart(dataset_with_name).transform_calculate(
    Season="'Summer Games'"
).mark_point(
    opacity=0,
    size=0,
    filled=True
).encode(
    shape=alt.Shape('Season:N').scale(
        domain=['Summer Games', 'Winter Games'],
        range=['circle', 'diamond']
    ).legend(title="Olympic Cycle"),
    size=alt.Size('Season:N').scale(
        domain=['Summer Games', 'Winter Games'],
        range=[175, 250]
    )
)

hover = alt.selection_point(
    fields=['Year'], nearest=True, on='mouseover', empty='none', clear='mouseout'
)

tooltips = alt.Chart(dataset_with_name).transform_pivot(
    'Country', value='GDP Growth Percentage', groupby=['Year']
).mark_rule(color='gray').encode(
    x='Year:T',
    opacity=alt.condition(hover, alt.value(1.0), alt.value(0.0)),
    tooltip=[alt.Tooltip('Year:T', title='Year', format='%Y')]
).add_params(hover)

summer_dots = alt.Chart(dataset_with_name).transform_filter(
    # Olympic Years Only
    summer_filter
).transform_calculate(
    # Create a real column named 'Season' containing the string literal
    Season="'Summer Games'",
    Formatted_GDP="format(datum['GDP Growth Percentage'], '.2f') + '%'",
    Formatted_GDP_5="format(datum['Avg GDP (>=5 Yrs Before)'], '.2f') + '%'",
    Formatted_GDP_10="format(datum['Avg GDP (>=10 Yrs After)'], '.2f') + '%'"
).mark_point(
    filled=True,
    opacity=1
).encode(
    x='Year:T',
    y='GDP Growth Percentage:Q',
    color='Country:N',
    size=alt.condition(selection, alt.value(275), alt.value(275)),
    opacity=alt.condition(selection, alt.value(1.0), alt.value(0.1)),
    tooltip=[
        alt.Tooltip('Year:T', format='%Y'),
        alt.Tooltip('Season:N'),
        alt.Tooltip('Country:N'),
        alt.Tooltip('Formatted_GDP:N', title="GDP Growth"),
        alt.Tooltip('TotalRevenue:N', title='Total Revenue'),
        alt.Tooltip('TotalCost:N', title='Total Cost'),
        alt.Tooltip('Formatted_GDP_5:N', title='Avg GDP (>=5 Yrs Before)'),
        alt.Tooltip('Formatted_GDP_10:N', title='Avg GDP (>=10 Yrs After)')
    ]
)

winter_dots = alt.Chart(dataset_with_name).transform_filter(
    # Olympic Years Only
    winter_filter
).transform_calculate(
    # Create a real column named 'Season' containing the string literal
    Season="'Winter Games'",
    Formatted_GDP="format(datum['GDP Growth Percentage'], '.2f') + '%'",
    Formatted_GDP_5="format(datum['Avg GDP (>=5 Yrs Before)'], '.2f') + '%'",
    Formatted_GDP_10="format(datum['Avg GDP (>=10 Yrs After)'], '.2f') + '%'"
).mark_point(
    filled=True,
    opacity=1,
    shape='diamond'
).encode(
    x='Year:T',
    y=alt.Y('GDP Growth Percentage:Q', title="% GDP Growth"),
    color='Country:N',
    size=alt.condition(selection, alt.value(350), alt.value(350)),
    opacity=alt.condition(selection, alt.value(1.0), alt.value(0.1)),
    tooltip=[
        alt.Tooltip('Year:T', format='%Y'),
        alt.Tooltip('Season:N'),
        alt.Tooltip('Country:N'),
        alt.Tooltip('Formatted_GDP:N', title="GDP Growth"),
        alt.Tooltip('TotalRevenue:N', title='Total Revenue'),
        alt.Tooltip('TotalCost:N', title='Total Cost'),
        alt.Tooltip('Formatted_GDP_5:N', title='Avg GDP (>=5 Yrs Before)'),
        alt.Tooltip('Formatted_GDP_10:N', title='Avg GDP (>=10 Yrs After)')
    ]
)

line_graph_gdp = (cycle_legend_anchor + tooltips + lines + summer_dots + winter_dots)
line_graph_gdp = line_graph_gdp.configure_legend(
    unselectedOpacity=0.1
)

line_graph_gdp.display()
```
